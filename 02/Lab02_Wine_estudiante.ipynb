## Mini-lab: ¿qué ve la máquina sin etiquetas?

### 1. Setup: recuperar Wine y quitarle las etiquetas

Mismo dataset de la semana pasada. La única diferencia: hoy `y` se guarda aparte y **no entra al
modelo**.

**Antes de ejecutar:** si te paso solo `X`, ¿cómo sabrías cuántos tipos de vino hay?

```python
from sklearn.datasets import load_wine
import pandas as pd

wine = load_wine(as_frame=True)
df = wine.frame

X = df.drop(columns="target")   # ← hoy trabajamos SIN el target
y = df["target"]                # lo guardamos solo para comparar al final

print("X:", X.shape, "— sin etiquetas")
X.head()
```

### 2. K-means sobre los vinos

Lo conoces de Minería de Datos. Hoy el foco no es el algoritmo, es el **paradigma**: mira la firma.

```python
from sklearn.cluster import KMeans

km = KMeans(n_clusters=3, n_init=10, random_state=42)
clusters = km.fit_predict(X)    # nótese: fit_predict SIN y

print("clusters encontrados:", pd.Series(clusters).value_counts().sort_index().to_dict())
```
